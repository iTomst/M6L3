```python
from imageai.Detection import ObjectDetection

# Instantiating the object detection class
detector = ObjectDetection()

# Setting a path to the YOLOv3 model
model_path = "/content/yolov3.pt"

# Installing the YOLOv3 model and setting a path to the weights file
detector.setModelTypeAsYOLOv3()
detector.setModelPath(model_path)

# Loading the model
detector.loadModel()
```

```python
# A list of all object the model knows about
detector.CustomObjects()
```

```python
# Reconocer imágenes en la imagen seleccionada con una probabilidad de al menos el 30% y guardarlas en un nuevo archivo.
detections = detector.detectObjectsFromImage(
    input_image="/content/Carretera 1.jpg",
    output_image_path="output_image.jpg",
    minimum_percentage_probability=30)
```

```python
for eachObject in detections:
    print(eachObject["name"] , " : ", eachObject["percentage_probability"], " : ", eachObject["box_points"] )
    print("--------------------------------")
```

```python
def detect_objects_on_road(input_image, output_image, model_path):
    from imageai.Detection import ObjectDetection

    detector = ObjectDetection()
    detector.setModelTypeAsYOLOv3()
    detector.setModelPath(model_path)
    detector.loadModel()

    detections = detector.detectObjectsFromImage(
        input_image=input_image,
        output_image_path=output_image,
        minimum_percentage_probability=30
    )

    return detections


def analyze_objects(detections):
    counts = {}

    for obj in detections:
        name = obj["name"]
        counts[name] = counts.get(name, 0) + 1

    print("--- Análisis de Objetos en la Vía ---")

    for name, count in counts.items():
        print(f"{name.capitalize()}: {count}")

    return counts


def road_safety_rules():
    print()
    print("Hola, soy SafetyAI, una aplicación de seguridad vial.")
    print("Las normas de seguridad vial son muy importantes.")
    print("Utiliza los semáforos y los pasos de peatones.")
    print("Nunca cruces una carretera por lugares que no estén diseñados para ello.")
    print("Al cruzar una carretera, debes ser prudente y prestar atención.")
    print("Ten cuidado en la carretera y ¡buena suerte!")


# Ejecutar el detector
detections = detect_objects_on_road(
    "/content/Carretera 1.jpg",
    "/content/output_image.jpg",
    "/content/yolov3.pt"
)

# Mostrar el análisis en texto
analyze_objects(detections)

# Mostrar las reglas de seguridad
road_safety_rules()

# Mostrar la imagen con los objetos detectados
from google.colab.patches import cv2_imshow
import cv2

img = cv2.imread("/content/output_image.jpg", cv2.IMREAD_UNCHANGED)
cv2_imshow(img)
```
